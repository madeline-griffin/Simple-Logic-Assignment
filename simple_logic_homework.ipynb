## Problem 1: Stacking logical operators

# 1.1 Add 1 to the value at index 300

```python
for i in range(len(influenza_genome)):
    if i == 300:
        influenza_genome[i] = influenza_genome[i] + 1

answer_1_1 = influenza_genome[300]
print( "1.1=" , answer_1_1)
```

# 1.2 First 30 elements: if divisible by 3, multiply the value by 3

```python
for i in range(30):
    if influenza_genome[i] % 3 == 0:
        influenza_genome[i] = influenza_genome[i] * 3

answer_1_2 = influenza_genome[:30]
print("1.2=", answer_1_2)
```

# 1.3 Last 30 elements: if the index is divisible by 5, replace the value with "a"

```python
n = len(influenza_genome)
for i in range(n - 30, n):
    if i % 5 == 0:
        influenza_genome[i] = "a"

answer_1_3 = influenza_genome[-30:]
print("1.3=", answer_1_3)
```

# 1.4 Elements between index 200 and 300: if divisible by both 3 and 5, replace with 10


```python
for i in range(200, 300):
    value = influenza_genome[i]
    if isinstance(value, int) and value % 3 == 0 and value % 5 == 0:
        influenza_genome[i] = 10

answer_1_4 = influenza_genome[200:300]
print("1.4=", answer_1_4)
```

# Problem 2: Loops within loops

# 2.1 For loop over indices 234 through 237

```python
answer_2_1 = []
for i in range(234, 238):
    print("2.1=", influenza_genome[i])
    answer_2_1.append(influenza_genome[i])
```

# 2.2 While loop over indices 234 through 237

```python
answer_2_2 = []
i = 234
while i <= 237:
    print("2.2=", influenza_genome[i])
    answer_2_2.append(influenza_genome[i])
    i += 1
```

# 2.3 For loop over indices 234 through 237; at index 236 print the item 7 times

```python
answer_2_3 = []
for i in range(234, 238):
    if i == 236:
        for _ in range(7):
            print("2.3=", influenza_genome[i])
            answer_2_3.append(influenza_genome[i])
    else:
        print("2.3=", influenza_genome[i])
        answer_2_3.append(influenza_genome[i])
```

# Problem 3: Functions

# 3.1 Ten values spaced 25 indices apart (0, 25, 50, ...)

```python
def spaced_by_25(genome):
    results = []
    for i in range(0, 250, 25):      # 0, 25, ..., 225 -> 10 values
        results.append(genome[i])
    print("3.1=", results)
    return results

answer_3_1 = spaced_by_25(influenza_genome)
```

# 3.2 Twenty values spaced `y` indices apart

```python
def spaced_by_y(genome, y=15):
    # 20 values with spacing y need index 19*y < len(genome); y = 15 gives a max index of 285
    if 19 * y >= len(genome):
        raise ValueError("y is too large for this dataset")
    results = []
    for i in range(20):
        results.append(genome[i * y])
    print("3.2=", results)
    return results

answer_3_2 = spaced_by_y(influenza_genome, y=15)
```

# 3.3 Print every other item from the output of 3.2 (10 outputs)

```python
def every_other(values):
    printed = []
    for i in range(0, len(values), 2):
        print("3.3=", values[i])
        printed.append(values[i])
    return printed

answer_3_3 = every_other(answer_3_2)
```

# Problem 4: Putting it all together


```python
def combined(genome):
    modified = list(genome)          # work on a copy; the input list is left unchanged

    # Logic from 1.4
    for i in range(200, 300):
        value = modified[i]
        if isinstance(value, int) and value % 3 == 0 and value % 5 == 0:
            modified[i] = 10

    # Logic from 2.3
    printed = []
    for i in range(234, 238):
        if i == 236:
            for _ in range(7):
                print("4.0=", modified[i])
                printed.append(modified[i])
        else:
            print("4.0=", modified[i])
            printed.append(modified[i])
    return printed

answer_4 = combined(original_genome)
```
